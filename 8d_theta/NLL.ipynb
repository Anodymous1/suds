```python
from object_handler import load_pickle
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Train your model as normal
inference = load_pickle("./model_5/inference_2048.pkl")

# Extract the training history from the SNLE object
train_losses = -np.array(inference._summary["training_log_probs"])
val_losses = -np.array(inference._summary["validation_log_probs"])

# Save the metrics to disk alongside the model
metrics = {
    "train_losses": train_losses,
    "val_losses": val_losses
}


# Plot the diagnostic curves using Matplotlib

plt.figure(figsize=(8, 5))
plt.plot(train_losses, label="Train Loss")
plt.plot(val_losses, label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Negative Log-Probability (Loss)")
plt.legend()
plt.title("NLE Training vs. Validation Loss")
plt.grid(True, linestyle="--", alpha=0.6)
# plt.savefig("./8d_theta/model_3/loss_curve.png")
plt.show()
```
