# Generating data for model 8

```python
import os
os.environ["OMP_NUM_THREADS"] = "2"     # agama
os.environ["MKL_NUM_THREADS"] = "2"     # numpy, scipy
os.environ["OPENBLAS_NUM_THREADS"] = "2"    # numpy
os.environ["NUMEXPR_NUM_THREADS"] = "1"     # pandas

import agama
import torch 
import numpy as np
from scipy import integrate
from astropy import units as u

from sbi.utils import BoxUniform
from sbi.inference import SNLE, simulate_for_sbi, prepare_for_sbi
from sbi.utils import likelihood_nn

from sklearn.metrics import mean_squared_error, r2_score

import pandas as pd
import pickle
import matplotlib.pyplot as plt
from galaxy_generation import generate_galaxy_multiple
from prior_generation import generate_prior


import corner

torch.set_num_threads(4)

# set agama unit to be in Msun, kpc, km/s
agama.setUnits(mass=1 * u.Msun, length=1*u.kpc, velocity=1 * u.km /u.s)

agama.setRandomSeed(13)
torch.manual_seed(13)
np.random.seed(13)
```

```python
num_galaxies = 100_000
prior = generate_prior()
thetas = prior.sample((num_galaxies,))
n_stars = np.random.poisson(100, size=num_galaxies)

theta = torch.repeat_interleave(thetas, torch.tensor(n_stars), dim=0)
x = generate_galaxy_multiple(thetas, n_stars, 4)

```

```python
pd.DataFrame(theta).to_csv("training_theta_g100k_s100p.csv", header=False, index=False)
pd.DataFrame(x).to_csv("training_x_g100k_s100p.csv", header=False, index=False)
```
